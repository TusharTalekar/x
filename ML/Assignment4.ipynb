```python
gdx = []
gdy = []

def gradient_descent(x,learning_rate,iterations):
    gdx.append(x)
    gdy.append((x+3)**2)
    for i in range(iterations):
        gradient = 2*(x+3)
        x = x- learning_rate*gradient

        gdx.append(x)
        gdy.append((x+3)**2)
        
        print(f"Iteration {i+1} => x = {x}, y = {(x+3)**2}")
    return x

x=2
lr=0.1
it=50

min_x = gradient_descent(x,lr,it)
print("Minima at: ")
print("x = ",min_x)
print("x = ",(min_x+3)**2)
```

```python
import seaborn as sns

sns.scatterplot(x=gdx,y=gdy)
```
